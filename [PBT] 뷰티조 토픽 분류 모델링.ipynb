## 5. 속성값이 모두 0인 행 개수 확인


```python
cols = [
    '보습', '향', '발림성', '자극', '용기',
    '배송', '기능', '가격'
]

# 9개 속성이 모두 0인 행 찾기
zero_rows = (df[cols] == 0).all(axis=1)

# 전체 개수 확인
print("속성값이 모두 0인 행 개수:", zero_rows.sum())

# 국가별 개수 확인
print("\n국가별 속성값 전체 0인 행 개수:")
display(df.loc[zero_rows, '국가'].value_counts())

# 해당 행 직접 확인
display(df[zero_rows])


# 한국만
kr_zero = df.loc[zero_rows]
kr_zero = kr_zero[kr_zero['국가'] == '한국']

with pd.option_context('display.max_rows', None, 'display.max_columns', None):
    display(kr_zero)

```

## 6. 속성값이 모두 0인 행 제거

```python
df1 = df.loc[~zero_rows].copy()

print("삭제 전 행 개수:", len(df))
print("삭제 후 행 개수:", len(df1))
print("삭제된 행 개수:", len(df) - len(df1))

display(df1.head())
```
